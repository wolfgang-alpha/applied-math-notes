## Strain and stress, a closer look

### Where the strain tensor comes from

Two neighbouring material points $x$ and $x + dx$ are displaced by $u(x)$ and by $u(x + dx) = u(x) + \nabla u \, dx$
(first order), so the connecting vector $dx$ becomes $dx + \nabla u \, dx$. The displacement gradient is the sum of a
symmetric and an antisymmetric part,

$$\nabla u = \varepsilon + \omega, \qquad \varepsilon = \tfrac12 \left(\nabla u + \nabla u^T\right), \qquad
  \omega = \tfrac12 \left(\nabla u - \nabla u^T\right).$$

The antisymmetric part is a **small rotation**: in 2D, $\omega = \begin{pmatrix} 0 & -\theta \\ \theta & 0 \end{pmatrix}$
with $\theta = \tfrac12 \left(\partial u_y / \partial x - \partial u_x / \partial y\right)$, and $\omega \, dx$ turns $dx$ by
the angle $\theta$ without changing its length (to first order). A rigid motion $u = c + \Omega x$ with constant
antisymmetric $\Omega$ therefore has $\varepsilon = 0$. The change of length of $dx$ is

$$|dx + \nabla u \, dx|^2 - |dx|^2 = 2 \, dx \cdot \varepsilon \, dx + O(|\nabla u|^2),$$

only the symmetric part enters: $\varepsilon$ is the **strain**. Its components have a direct meaning:
$\varepsilon_{xx}$ is the relative elongation of a fibre in $x$ direction, and $2 \varepsilon_{xy} = \gamma_{xy}$ is the
decrease of the right angle between fibres in $x$ and $y$ direction (shear angle). $\operatorname{tr}\varepsilon$ and
$\operatorname{dev}\varepsilon$ are the change of volume and the change of shape of lecture 2.

### Why the stress tensor is symmetric

Two independent arguments, one from the energy and one from mechanics.

- **Energy.** A rigid rotation stores no energy, so the strain energy can depend on $\nabla u$ only through its
  symmetric part, $\psi = \psi(\varepsilon)$: the material model of lecture 2 is a function of $\operatorname{tr}\varepsilon$
  and $\operatorname{dev}\varepsilon$ and never sees $\omega$. The stress $\sigma = \partial\psi / \partial\varepsilon$, the
  derivative with respect to a symmetric tensor, is then symmetric itself.
- **Moment equilibrium** (balance of angular momentum, "Momentengleichgewicht"), independent of the material.
  Consider a small cube with the edges $dx, dy, dz$. The shear stresses $\sigma_{xy}$ on the two faces with normal
  $\pm x$ form a couple with the moment $\sigma_{xy} \, (dy \, dz) \, dx$ about the $z$ axis, the shear stresses
  $\sigma_{yx}$ on the faces with normal $\pm y$ the opposite couple $\sigma_{yx} \, (dx \, dz) \, dy$. Normal stresses,
  body forces and the variation of the stresses over the cube contribute moments of higher order in the edge
  length, so equilibrium of moments requires
  $$\left(\sigma_{xy} - \sigma_{yx}\right) dx \, dy \, dz = 0 \qquad \Longrightarrow \qquad \sigma_{xy} = \sigma_{yx},$$
  and likewise about the other two axes: $\sigma = \sigma^T$ (Cauchy's second law). Together with the force
  equilibrium $-\nabla \cdot \sigma = f$ of lecture 2 these are the equilibrium conditions of a continuum.

A symmetric strain thus produces a symmetric stress, and a symmetric stress is what mechanics demands anyway.
Symmetric tensors have real eigenvalues and orthogonal eigenvectors, which is what the next section is about.

### Principal stresses and invariants

The traction (force per area) on a plane with normal $n$ is $t = \sigma \, n$. On the planes where it is purely
normal, $\sigma \, n = \lambda \, n$, there is no shear stress: these are the **principal planes**, $n$ the principal
directions and the eigenvalues $\sigma_1 \ge \sigma_2 \ge \sigma_3$ the **principal stresses**. They are the roots of
the characteristic polynomial

$$\det(\sigma - \lambda I) = -\lambda^3 + I_1 \, \lambda^2 - I_2 \, \lambda + I_3 = 0,$$

$$I_1 = \operatorname{tr}\sigma = \sigma_1 + \sigma_2 + \sigma_3, \qquad
  I_2 = \tfrac12 \big[(\operatorname{tr}\sigma)^2 - \operatorname{tr}(\sigma^2)\big] = \sigma_1 \sigma_2 + \sigma_2 \sigma_3 + \sigma_3 \sigma_1, \qquad
  I_3 = \det\sigma = \sigma_1 \sigma_2 \sigma_3 .$$

The stress components change when the coordinate system is rotated, the eigenvalues do not, and so $I_1, I_2, I_3$
are the same in every coordinate system: the **invariants** of $\sigma$. As in lecture 2 the stress is split into
a hydrostatic (mean) part and a deviator,

$$p = \tfrac13 I_1, \qquad s = \sigma - p \, I, \qquad \operatorname{tr} s = 0,$$

and the invariants of the deviator are

$$J_2 = \tfrac12 \, s : s = \tfrac16 \big[(\sigma_1 - \sigma_2)^2 + (\sigma_2 - \sigma_3)^2 + (\sigma_3 - \sigma_1)^2\big],
  \qquad J_3 = \det s .$$

$J_2$ measures how far the stress state is from a purely hydrostatic one. In two dimensions the eigenvalues have the
closed form (Mohr's circle)

$$\sigma_{a, b} = \frac{\sigma_{xx} + \sigma_{yy}}{2} \pm \sqrt{\left(\frac{\sigma_{xx} - \sigma_{yy}}{2}\right)^2 + \sigma_{xy}^2},$$

and in plane strain the third principal stress is $\sigma_{zz} = \lambda \, \operatorname{tr}\varepsilon = \nu \, (\sigma_{xx} + \sigma_{yy})$,
which is not zero and has to be included.

The same decomposition applies to the strain, and it is exactly the one behind the strain energy of lecture 2:
with $\sigma = K \, (\operatorname{tr}\varepsilon) \, I + 2G \, \operatorname{dev}\varepsilon$ one finds
$I_1 = 3K \operatorname{tr}\varepsilon$ and $s = 2G \, \operatorname{dev}\varepsilon$, hence

$$\psi = \tfrac12 K \, (\operatorname{tr}\varepsilon)^2 + G \, \operatorname{dev}\varepsilon : \operatorname{dev}\varepsilon
  = \frac{I_1^2}{18 K} + \frac{J_2}{2 G} :$$

the volumetric energy depends on $I_1$ only, the distortion energy on $J_2$ only.

### Failure theories as functions of the invariants

The tension test of lecture 2 tells us the stress at which a material yields ($\sigma_y$, ductile) or breaks
($\sigma_f$, brittle) in one particular stress state. A **failure theory** extends this to arbitrary states through an
**equivalent stress** $\sigma_{eq}(\sigma)$ that reduces to $\sigma_{xx}$ in the tension test: the material fails when
$\sigma_{eq}$ reaches the measured value. An isotropic material has no preferred direction, so $\sigma_{eq}$ can only
depend on the invariants, or equivalently on the principal stresses. Three classical theories, in the order
brittle, ductile, very ductile:

- **Rankine, maximum normal stress** (brittle materials: cast iron, concrete, glass, ceramics). They break by
  separation perpendicular to the largest tensile stress:
  $$\sigma_R = \sigma_1 \le \sigma_f \qquad (\text{and } |\sigma_3| \le \sigma_{f,c} \text{ for the usually much higher compressive strength}).$$
  It depends on the hydrostatic part of the stress.
- **Tresca, maximum shear stress** (ductile materials). Metals yield by slip on the planes of maximum shear
  stress, which lie at $45^\circ$ between the largest and the smallest principal direction,
  $\tau_{max} = \tfrac12 (\sigma_1 - \sigma_3)$. In the tension test $\tau_{max} = \sigma_{xx} / 2$, so
  $$\sigma_T = \sigma_1 - \sigma_3 \le \sigma_y .$$
  A hydrostatic pressure shifts all three principal stresses equally and leaves $\sigma_T$ unchanged: it does not
  cause yielding.
- **von Mises, distortion energy** (very ductile materials such as mild steel and aluminium alloys). Yielding starts
  when the distortion energy $J_2 / (2G)$ reaches the value it has in the tension test at yield, $J_2 = \sigma_y^2 / 3$:
  $$\sigma_{vM} = \sqrt{3 J_2} = \sqrt{\tfrac12 \big[(\sigma_1 - \sigma_2)^2 + (\sigma_2 - \sigma_3)^2 + (\sigma_3 - \sigma_1)^2\big]} \le \sigma_y .$$
  Like Tresca it ignores the hydrostatic part, but it is smooth and takes the intermediate principal stress into
  account; it is the most widely used criterion for metals.

Two limiting cases. In **pure shear** ($\sigma_1 = \tau$, $\sigma_2 = 0$, $\sigma_3 = -\tau$) the criteria give
$\sigma_T = 2\tau$ and $\sigma_{vM} = \sqrt{3} \, \tau$: Tresca predicts yielding at $\tau = \sigma_y / 2$, von Mises at
$\tau = \sigma_y / \sqrt{3} = 0.577 \, \sigma_y$, a difference of 15 % with Tresca on the safe side; this is the largest
difference between the two. Under **hydrostatic tension** ($\sigma_1 = \sigma_2 = \sigma_3 = p$) both give $0$, no
yielding at any $p$, whereas Rankine gives $\sigma_R = p$: brittle fracture is the failure mode the ductile criteria
cannot see.

The figure below shows the three failure surfaces in the plane $\sigma_3 = 0$: a square (Rankine), a hexagon (Tresca)
and an ellipse (von Mises), all through the tension test point $(\sigma_y, 0)$. One more remark for the FEniCS examples:
the plane-strain problems of lecture 2 always carry $\sigma_{zz} = \nu \, (\sigma_{xx} + \sigma_{yy}) \ne 0$, and the
helper functions below include it in the principal stresses.

```python
from fenics import *
from matplotlib import pyplot
import numpy as np
%matplotlib inline

set_log_level(LogLevel.WARNING)   # silence the solver messages
```

```python
# failure surfaces in the plane sigma_3 = 0, normalized with sigma_y = sigma_f = 1
s1, s2 = np.meshgrid(np.linspace(-1.5, 1.5, 601), np.linspace(-1.5, 1.5, 601))
rankine_   = np.maximum(abs(s1), abs(s2))                 # |sigma_1|, |sigma_2| <= sigma_f
tresca_    = np.maximum(rankine_, abs(s1 - s2))           # sigma_1 - sigma_3 with sigma_3 = 0 as the third value
von_mises_ = np.sqrt(s1**2 - s1*s2 + s2**2)               # sqrt(3 J_2)

pyplot.figure(figsize=(7, 5), dpi=100)
for f, colour, name in ((rankine_, 'C2', 'Rankine'), (tresca_, 'C1', 'Tresca'), (von_mises_, 'C0', 'von Mises')):
    pyplot.contour(s1, s2, f, levels=[1.0], colors=colour)
    pyplot.plot([], [], color=colour, label=name)
pyplot.plot([1, 0, -1, 0], [0, 1, 0, -1], 'ko', label='tension / compression test')
pyplot.plot([0.5, -0.5], [-0.5, 0.5], 'k^', label='pure shear, Tresca')
pyplot.plot([1/np.sqrt(3), -1/np.sqrt(3)], [-1/np.sqrt(3), 1/np.sqrt(3)], 'ks', label='pure shear, von Mises')
pyplot.axis('equal')
pyplot.xlabel(r'$\sigma_1 / \sigma_y$')
pyplot.ylabel(r'$\sigma_2 / \sigma_y$')
pyplot.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0))
pyplot.savefig("stress-0.png", bbox_inches='tight')
```

## Equivalent stresses in FEniCS

The setup of lecture 2 is reused: plane strain, $E = 1000$, $\nu = 0.3$, the strain energy in terms of $K$ and $G$,
the stress by differentiation of the energy, and the total potential energy as the object to minimize. New are the
helper functions for the principal stresses (2D closed form plus $\sigma_{zz}$) and for the three equivalent stresses,
and `show_stresses`, which plots them side by side. Because we now look at stresses, i.e. at derivatives of the
displacement, **quadratic** Lagrange elements are used: the stress is then linear within each element instead of
constant, and both the pictures and the point values become much better.

```python
tol = 1e-3
E, nu = 1000.0, 0.3
K = E/(3*(1 - 2*nu))
G = E/(2*(1 + nu))

def epsilon(u):
    return sym(grad(u))

def psi(eps):
    # strain energy density, volumetric + deviatoric part (3D deviator, plane strain)
    return 0.5*K*tr(eps)**2 + G*(inner(eps, eps) - tr(eps)**2/3)

def sigma(u):
    # stress = derivative of the strain energy with respect to the strain
    eps = variable(epsilon(u))
    return diff(psi(eps), eps)

def principal_stresses(u):
    # in-plane principal stresses (Mohr's circle) and the out-of-plane stress of plane strain
    s = sigma(u)
    c = 0.5*(s[0, 0] + s[1, 1])
    r = sqrt(0.25*(s[0, 0] - s[1, 1])**2 + s[0, 1]**2)
    s_zz = (K - 2*G/3)*tr(epsilon(u))          # lambda * tr(eps)
    return c + r, c - r, s_zz

def larger(a, b):
    return 0.5*(a + b + abs(a - b))

def smaller(a, b):
    return 0.5*(a + b - abs(a - b))

def sigma_1(u):
    s_a, s_b, s_zz = principal_stresses(u)      # s_a >= s_b always
    return larger(s_a, s_zz)

def sigma_3(u):
    s_a, s_b, s_zz = principal_stresses(u)
    return smaller(s_b, s_zz)

def rankine(u):
    return sigma_1(u)

def tresca(u):
    return sigma_1(u) - sigma_3(u)

def von_mises(u):
    s_a, s_b, s_zz = principal_stresses(u)
    return sqrt(0.5*((s_a - s_b)**2 + (s_b - s_zz)**2 + (s_zz - s_a)**2))

def left(x, on_boundary):
    return on_boundary and near(x[0], 0.0, tol)

def solve_elasticity(V, bcs, f=Constant((0.0, 0.0)), T=None, ds_T=None):
    # minimization of the total potential energy as in lecture 2; T is a traction on the boundary part ds_T
    u = Function(V)
    v = TestFunction(V)
    Pi = ( psi(epsilon(u)) - dot(f, u) )*dx
    if T is not None:
        Pi = Pi - dot(T, u)*ds_T
    solve(derivative(Pi, u, v) == 0, u, bcs)
    return u

def show_stresses(u, filename, layout=(3, 1), figsize=(8, 7), zoom=None):
    # the three equivalent stresses side by side (projected to P1 by plot)
    pyplot.figure(figsize=figsize, dpi=100)
    for i, (f, name) in enumerate(((von_mises, 'von Mises'), (tresca, 'Tresca'), (rankine, 'Rankine (largest principal stress)'))):
        pyplot.subplot(*layout, i + 1)
        c = plot(f(u), title=name)
        pyplot.colorbar(c, fraction=0.046, pad=0.04)
        if zoom is not None:
            pyplot.xlim(0, zoom)
            pyplot.ylim(0, zoom)
    pyplot.tight_layout()
    pyplot.savefig(filename)
```

### 1. Tension test

The beam $(0, 5) \times (0, 1)$ with the left end fixed and the right end pulled to $u = (0.25, 0)$, example 3 of
lecture 2. Away from the ends the state is uniaxial plane strain: $\varepsilon_{xx} = 0.05$, $\sigma_{yy} = 0$,
$\sigma_{xx} = E' \varepsilon_{xx}$ with $E' = E / (1 - \nu^2)$, and $\sigma_{zz} = \nu \, \sigma_{xx}$. With the principal
stresses $(\sigma_{xx}, 0, \nu \sigma_{xx})$ Tresca and Rankine give $\sigma_{xx}$, von Mises gives
$\sigma_{xx} \sqrt{1 - \nu + \nu^2} = 0.889 \, \sigma_{xx}$: the constraint in $z$ direction adds a tensile $\sigma_{zz}$,
which raises the hydrostatic part and lowers the distortion energy. The values at the centre of the bar are
compared with these numbers (they agree to about 1 %, the ends being held in $y$ direction as well); the pictures
show the stress concentrations at the corners.

```python
length, height = 5.0, 1.0
mesh = RectangleMesh(Point(0.0, 0.0), Point(length, height), 50, 10)
V    = VectorFunctionSpace(mesh, 'Lagrange', 2)
W    = FunctionSpace(mesh, 'Lagrange', 1)          # for point values of stresses (projection)

def right(x, on_boundary):
    return on_boundary and near(x[0], length, tol)

# mark the right end with 1, so that ds(1) integrates over it only
boundaries = MeshFunction('size_t', mesh, mesh.topology().dim() - 1, 0)
AutoSubDomain(right).mark(boundaries, 1)
ds = Measure('ds', domain=mesh, subdomain_data=boundaries)

u = solve_elasticity(V, [DirichletBC(V, Constant((0.0, 0.0)), left), DirichletBC(V, Constant((0.25, 0.0)), right)])

s_xx   = E/(1 - nu**2)*0.05
centre = (length/2, height/2)
print("sigma_xx  %.2f  (uniaxial plane strain: %.2f)" % (project(sigma(u)[0, 0], W)(*centre), s_xx))
print("von Mises %.2f  (%.2f)" % (project(von_mises(u), W)(*centre), s_xx*np.sqrt(1 - nu + nu**2)))
print("Tresca    %.2f  (%.2f)" % (project(tresca(u), W)(*centre), s_xx))
print("Rankine   %.2f  (%.2f)" % (project(rankine(u), W)(*centre), s_xx))
show_stresses(u, "stress-1.png")
```

### 2. Cantilever with an end load

Example 5 of lecture 2: clamped at the left end, traction $T = (0, -0.5)$ on the right end, no body force. Beam
theory gives at the cross section $x$ the bending moment $M = P \, (L - x)$ with $P = 0.5 \, H$ and the shear force $P$,
and from them the linear bending stress and the parabolic shear stress over the height,

$$\sigma_{xx} = \frac{M \, (y - H/2)}{I}, \qquad
  \sigma_{xy} = -\frac{3 P}{2 H} \left[1 - \left(\frac{y - H/2}{H/2}\right)^2\right], \qquad I = \frac{H^3}{12}.$$

The 2D stresses at $x = 2.5$ are compared with these formulas. The equivalent stresses are largest in the top and
bottom fibres near the clamped end, where the bending moment is largest, and small along the neutral axis, where
only the shear stress acts.

```python
u = solve_elasticity(V, [DirichletBC(V, Constant((0.0, 0.0)), left)], T=Constant((0.0, -0.5)), ds_T=ds(1))

P, I  = 0.5*height, height**3/12
x_cut = 2.5
yy    = np.linspace(0.0, height, 11)
sxx, sxy = project(sigma(u)[0, 0], W), project(sigma(u)[0, 1], W)
s_xx = np.array([sxx(x_cut, y) for y in yy])
s_xy = np.array([sxy(x_cut, y) for y in yy])
s_xx_beam = P*(length - x_cut)*(yy - height/2)/I
s_xy_beam = -1.5*P/height*(1 - ((yy - height/2)/(height/2))**2)
print("largest deviation from beam theory at x = %.1f:  sigma_xx %.3f (of %.2f),  sigma_xy %.3f (of %.2f)"
      % (x_cut, abs(s_xx - s_xx_beam).max(), abs(s_xx_beam).max(), abs(s_xy - s_xy_beam).max(), abs(s_xy_beam).max()))

pyplot.figure(figsize=(8, 3.5), dpi=100)
pyplot.subplot(1, 2, 1)
pyplot.plot(s_xx, yy, 'o', label='2D FEM')
pyplot.plot(s_xx_beam, yy, label='beam theory')
pyplot.xlabel(r'$\sigma_{xx}$ at x = 2.5')
pyplot.ylabel('y')
pyplot.legend()
pyplot.subplot(1, 2, 2)
pyplot.plot(s_xy, yy, 'o', label='2D FEM')
pyplot.plot(s_xy_beam, yy, label='beam theory')
pyplot.xlabel(r'$\sigma_{xy}$ at x = 2.5')
pyplot.legend()
pyplot.tight_layout()
pyplot.savefig("stress-2.png")

show_stresses(u, "stress-3.png")
```

Example 4 of lecture 2, $f = (0, -\rho g)$ with $\rho g = 0.3$. The bending moment now grows quadratically towards the
clamped end, $M(x) = \tfrac12 q \, (L - x)^2$ with $q = \rho g H$. Directly at the clamped corners the 2D solution is

```python
rho_g = 0.3
u = solve_elasticity(V, [DirichletBC(V, Constant((0.0, 0.0)), left)], f=Constant((0.0, -rho_g)))

q = rho_g*height
M = 0.5*q*(length - 1.0)**2
print("sigma_xx at (1, H): 2D %.2f, beam theory M (H/2) / I = %.2f" % (project(sigma(u)[0, 0], W)(1.0, height), M*(height/2)/I))
show_stresses(u, "stress-4.png")
```

### 4. Plate with a circular hole

A different body and the classical example of a stress concentration: a large plate with a hole of radius $a$ under
the far-field tension $\sigma_\infty$ in $x$ direction. Kirsch's solution (1898) gives on the edge of the hole the
hoop stress

$$\sigma_{\theta\theta}(a, \theta) = \sigma_\infty \, (1 - 2 \cos 2\theta),$$

i.e. $3 \, \sigma_\infty$ at the top and the bottom of the hole ($\theta = \pm 90^\circ$, stress concentration factor 3)
and $-\sigma_\infty$ at the sides. The radial and the shear stress vanish on the edge, so the principal stresses
there are $(\sigma_{\theta\theta}, 0, \nu \, \sigma_{\theta\theta})$.

The model: one quarter of the plate, by symmetry $u_x = 0$ on $x = 0$ and $u_y = 0$ on $y = 0$, cut out as a quarter
disc of radius $20 a$ around the hole; on its outer arc the traction $\sigma_\infty \, n_x \, e_x$ of the uniform stress
state is applied, the hole is traction free. The mesh is a rectangle in $(r, \theta)$ mapped to
$(x, y) = (r \cos\theta, r \sin\theta)$, as for the curved beam of the `linear-elasticity` notebook, and graded towards
the hole, where the stress varies quickly.

```python
a, R = 1.0, 20.0        # radius of the hole, outer radius of the model
sigma_inf = 1.0

mesh_h = RectangleMesh(Point(0.0, 0.0), Point(1.0, np.pi/2), 30, 30)
s, theta = mesh_h.coordinates().T.copy()
r = a + (R - a)*s**2                                   # graded: fine near the hole
mesh_h.coordinates()[:, 0] = r*np.cos(theta)
mesh_h.coordinates()[:, 1] = r*np.sin(theta)

V_h = VectorFunctionSpace(mesh_h, 'Lagrange', 2)
W_h = FunctionSpace(mesh_h, 'Lagrange', 1)

boundaries_h = MeshFunction('size_t', mesh_h, mesh_h.topology().dim() - 1, 0)
AutoSubDomain(lambda x, on_boundary: on_boundary and near(np.hypot(x[0], x[1]), R, 1e-2)).mark(boundaries_h, 1)
ds_h = Measure('ds', domain=mesh_h, subdomain_data=boundaries_h)
n = FacetNormal(mesh_h)

bcs = [DirichletBC(V_h.sub(0), Constant(0.0), lambda x, on_boundary: on_boundary and near(x[0], 0.0, tol)),   # symmetry
       DirichletBC(V_h.sub(1), Constant(0.0), lambda x, on_boundary: on_boundary and near(x[1], 0.0, tol))]
u = solve_elasticity(V_h, bcs, T=as_vector((sigma_inf*n[0], 0.0)), ds_T=ds_h(1))

print("hoop stress on the edge of the hole:  top   sigma_xx(0, a) = %.3f  (Kirsch: 3 sigma_inf = %.1f)"
      % (project(sigma(u)[0, 0], W_h)(0.0, a), 3*sigma_inf))
print("                                      side  sigma_yy(a, 0) = %.3f  (Kirsch: -sigma_inf = %.1f)"
      % (project(sigma(u)[1, 1], W_h)(a, 0.0), -sigma_inf))
print("maximum von Mises %.3f  (edge state (3, 0, 3 nu) sigma_inf: %.3f),  maximum Tresca %.3f  (%.1f)"
      % (project(von_mises(u), W_h).vector().max(), 3*sigma_inf*np.sqrt(1 - nu + nu**2),
         project(tresca(u), W_h).vector().max(), 3*sigma_inf))
show_stresses(u, "stress-5.png", layout=(1, 3), figsize=(12, 4), zoom=4.0)
```

## Buckling: stability as an eigenvalue problem

A slender column under compression does not fail because the stress reaches $\sigma_y$: long before that it
**buckles**, it suddenly bends sideways although the load is purely axial. This is a failure of the *stability* of the
straight equilibrium state, and the energy of lecture 2 explains it.

Consider the beam of lecture 2 under a compressive axial force $P$ and imagine a small lateral deflection $u(x)$.
Its bending energy is $\int \tfrac12 EI \, (u'')^2 \, dx$. The deflection also moves the loaded end towards the support:
a fibre element of length $dx$ inclined by the slope $u'$ covers only $dx \cos u' \approx dx \, (1 - \tfrac12 (u')^2)$ in
axial direction, so the end approaches by $\Delta = \int \tfrac12 (u')^2 \, dx$ and the load does the work $P \, \Delta$.
The total potential energy of the deflected state relative to the straight one is therefore

$$\Pi(u) = \int_0^L \tfrac12 \, EI \, (u'')^2 \, dx - \tfrac12 \, P \int_0^L (u')^2 \, dx .$$

For small $P$ the first term wins, $\Pi > 0$ for every deflection, and the straight state is a stable minimum. The
**critical load** is the smallest $P$ for which some deflection shape makes $\Pi = 0$: beyond it the load gains more
energy from the shortening than the beam stores in bending, and the straight column is no longer a minimum.
Stationarity of $\Pi$ in the direction $v$ (lecture 2) gives

$$\int_0^L EI \, u'' \, v'' \, dx - P \int_0^L u' \, v' \, dx = 0 \quad \text{for all } v
  \qquad \Longrightarrow \qquad EI \, u'''' + P \, u'' = 0,$$

a homogeneous equation whose only solution is $u = 0$, except for particular values of $P$ where nontrivial
solutions, the **buckling modes**, exist: an **eigenvalue problem**. With the finite element approximation
$u_h = \sum_i U_i \varphi_i$ it becomes the generalized matrix eigenvalue problem

$$K \, \phi = P \, K_G \, \phi, \qquad K_{ij} = \int_0^L EI \, \varphi_i'' \, \varphi_j'' \, dx, \qquad
  (K_G)_{ij} = \int_0^L \varphi_i' \, \varphi_j' \, dx,$$

with the stiffness matrix of lecture 2 and the **geometric stiffness** $K_G$ that comes from the work of the axial
load. Its smallest eigenvalue is the critical load and the eigenvector the buckling mode; equivalently
$P_{cr} = \min_u \int EI \, (u'')^2 \, dx \, / \int (u')^2 \, dx$, the Rayleigh quotient of the two energies. The same
kind of eigenvalue problem, with the mass matrix in place of the geometric stiffness, gives the vibration modes of
a body in lecture 4.

For the classical support cases Euler (1744) solved $EI \, u'''' + P \, u'' = 0$ by hand. The results are written as
$P_{cr} = \pi^2 EI / (K L)^2$ with the **effective length** $K L$, the distance between two points of zero bending
moment of the mode:

| support | $P_{cr}$ | $K$ |
|---|---|---|
| clamped – free (cantilever) | $\pi^2 EI / (4 L^2)$ | 2 |
| pinned – pinned | $\pi^2 EI / L^2$ | 1 |
| clamped – pinned | $2.046 \, \pi^2 EI / L^2$ | 0.699 |
| clamped – clamped | $4 \pi^2 EI / L^2$ | 0.5 |

### 5. Euler's column

The mixed formulation of lecture 2 is reused: deflection $u$ and slope $w = u'$ piecewise linear, the constraint
$w = u'$ enforced by the piecewise constant multiplier $p$. The bending energy is $\int \tfrac12 EI \, (w')^2 \, dx$ and the
work of the load $\tfrac12 P \int w^2 \, dx$, so the geometric stiffness is the "mass matrix" of the slope, $\int w \, z \, dx$.
Both matrices are assembled with `assemble`, the rows and columns of the Dirichlet degrees of freedom are removed,
and the dense generalized eigenvalue problem is solved with `scipy.linalg.eig` (the problems are tiny). The
constraint equations have no geometric stiffness and produce infinite eigenvalues, which are discarded.

Support conditions: a clamped end has $u = 0$ and $w = 0$, a pinned end only $u = 0$, a free end nothing; the
conditions on the bending moment and the shear force are natural, as in lecture 2.

```python
import scipy.linalg

def smallest_eigenpairs(K_mat, KG_mat, free, nr_modes):
    # K phi = lambda K_G phi on the free degrees of freedom; finite positive eigenvalues in increasing order
    lam, phi = scipy.linalg.eig(K_mat[np.ix_(free, free)], KG_mat[np.ix_(free, free)])
    ok = np.isfinite(lam) & (abs(lam.imag) <= 1e-8*abs(lam)) & (lam.real > 1e-8)
    lam, phi = lam[ok].real, phi[:, ok].real
    order = np.argsort(lam)[:nr_modes]
    return lam[order], phi[:, order]

def normalized(mode):
    # largest deflection 1
    values = mode.vector().get_local()
    mode.vector().set_local(values/values[np.argmax(abs(values))])
    return mode

def buckling_1d(length, nr_elements, EI, bcs_fn, nr_modes=3):
    mesh = IntervalMesh(nr_elements, 0.0, length)
    P1   = FiniteElement('Lagrange', mesh.ufl_cell(), 1)
    P0   = FiniteElement('DG', mesh.ufl_cell(), 0)
    V    = FunctionSpace(mesh, MixedElement(P1, P1, P0))
    (u, w, p) = TrialFunctions(V)
    (v, z, r) = TestFunctions(V)
    k   = ( EI*w.dx(0)*z.dx(0) + (w - u.dx(0))*r + (z - v.dx(0))*p )*dx    # bending energy + constraint
    k_g = w*z*dx                                                            # geometric stiffness
    K_mat, KG_mat = assemble(k).array(), assemble(k_g).array()
    fixed = set()
    for bc in bcs_fn(V):
        fixed.update(bc.get_boundary_values().keys())
    free = np.array(sorted(set(range(V.dim())) - fixed))
    P, phi = smallest_eigenpairs(K_mat, KG_mat, free, nr_modes)
    modes = []
    for j in range(len(P)):
        values = np.zeros(V.dim())
        values[free] = phi[:, j]
        mode = Function(V)
        mode.vector().set_local(values)
        modes.append(normalized(mode.split(deepcopy=True)[0]))       # deflection u of the mode
    return P, modes

EI, length = 1.0, 1.0

def right(x, on_boundary):
    return on_boundary and near(x[0], length, tol)

u_fixed = lambda V, end: DirichletBC(V.sub(0), Constant(0.0), end)     # u = 0 at the end
w_fixed = lambda V, end: DirichletBC(V.sub(1), Constant(0.0), end)     # u' = 0 at the end
supports = {
    'clamped - free':    (lambda V: [u_fixed(V, left), w_fixed(V, left)],                                        0.25),
    'pinned - pinned':   (lambda V: [u_fixed(V, left), u_fixed(V, right)],                                       1.0),
    'clamped - clamped': (lambda V: [u_fixed(V, left), w_fixed(V, left), u_fixed(V, right), w_fixed(V, right)],  4.0),
}

P_unit = np.pi**2*EI/length**2
xx = np.linspace(0.0, length, 201)
pyplot.figure(dpi=100)
for name, (bcs_fn, factor) in supports.items():
    P, modes = buckling_1d(length, 40, EI, bcs_fn)
    print("%-18s P_cr = %.4f = %.4f pi^2 EI/L^2   (Euler: %.2f)   higher modes: %.2f, %.2f pi^2 EI/L^2"
          % (name, P[0], P[0]/P_unit, factor, P[1]/P_unit, P[2]/P_unit))
    pyplot.plot(xx, [modes[0](x) for x in xx], label=name)
pyplot.xlabel("x")
pyplot.title("first buckling modes (largest deflection 1)")
pyplot.legend()
pyplot.savefig("buckling-0.png")
```

### 6. Buckling of a column in 2D

The same analysis for a two-dimensional body, without any beam theory. The slender column $(0, 20) \times (0, 1)$ is
clamped at the left end and compressed by the traction $T = (-1, 0)$ on the right end. Two steps:

1. **Prestress.** The linear problem of lecture 2 is solved for the unit load; its stress $\sigma_0$ is the (essentially
   uniform, compressive) state whose stability is in question.
2. **Geometric stiffness.** The second-order work of the prestress on a small additional displacement $u$ is
   $\tfrac12 \int_\Omega \sigma_0 : (\nabla u^T \nabla u) \; dx$, the 2D counterpart of $\tfrac12 P \int (u')^2 \, dx$: for
   $\sigma_0 = -\sigma \, e_x \otimes e_x$ it reduces to $-\tfrac12 \sigma \int |\partial_x u|^2 \, dx$. Stationarity gives
   $K \phi = \lambda \, K_G \phi$ with the stiffness matrix of lecture 2 and
   $(K_G)_{ij} = -\int_\Omega \sigma_0 : (\nabla\varphi_i^T \nabla\varphi_j) \; dx$; the smallest eigenvalue $\lambda$ is the
   factor by which the unit load has to be multiplied, $P_{cr} = \lambda \, H$.

The result is compared with Euler's cantilever load $\pi^2 E' I / (4 L^2)$, $E' = E / (1 - \nu^2)$, $I = H^3 / 12$, and
the second eigenvalue with the second Euler mode, $9 \pi^2 E' I / (4 L^2)$. Quadratic elements are essential here:
linear triangles are far too stiff in bending and overestimate the critical load by more than 20 %.

```python
length_c, height_c = 20.0, 1.0
mesh_c = RectangleMesh(Point(0.0, 0.0), Point(length_c, height_c), 40, 2)
V_c = VectorFunctionSpace(mesh_c, 'Lagrange', 2)

boundaries_c = MeshFunction('size_t', mesh_c, mesh_c.topology().dim() - 1, 0)
AutoSubDomain(lambda x, on_boundary: on_boundary and near(x[0], length_c, tol)).mark(boundaries_c, 1)
ds_c = Measure('ds', domain=mesh_c, subdomain_data=boundaries_c)
bc_c = DirichletBC(V_c, Constant((0.0, 0.0)), left)

# 1. prestress from the unit compressive load
u0 = solve_elasticity(V_c, [bc_c], T=Constant((-1.0, 0.0)), ds_T=ds_c(1))
sigma_0 = sigma(u0)

# 2. stiffness, geometric stiffness, eigenvalue problem
u = TrialFunction(V_c)
v = TestFunction(V_c)
k   = inner(sigma(u), epsilon(v))*dx
k_g = -inner(dot(grad(u), sigma_0), grad(v))*dx
K_mat, KG_mat = assemble(k).array(), assemble(k_g).array()
free = np.array(sorted(set(range(V_c.dim())) - set(bc_c.get_boundary_values().keys())))
lam, phi = smallest_eigenpairs(K_mat, KG_mat, free, 2)

P_euler = np.pi**2*E/(1 - nu**2)*height_c**3/12/(4*length_c**2)
print("critical load: 2D %.4f, Euler pi^2 E'I/(4L^2) = %.4f;   second mode: 2D %.3f, Euler %.3f"
      % (lam[0]*height_c, P_euler, lam[1]*height_c, 9*P_euler))

values = np.zeros(V_c.dim())
values[free] = phi[:, 0]
mode = Function(V_c)
mode.vector().set_local(values)
pyplot.figure(figsize=(8, 2.5), dpi=100)
plot(normalized(mode), mode='displacement', title='first buckling mode of the 2D column')
pyplot.savefig("buckling-1.png")
```
