## From the variational problem to the weak and the strong form

### The principle of minimum potential energy

A body in equilibrium takes the state $u$ that **minimizes its total potential energy**

$$\Pi(u) = \Psi(u) - W(u)$$

among all admissible $u$ (those that satisfy the essential boundary conditions). $\Psi$ is the strain energy, $W$ the work of the applied loads.

This minimization works for symmetric and positive operators $A$ with strong form $Au = f$. If $A$ is also **linear**, the weak form $a(u, v) = L(v)$ for all $v$ consists of a symmetric bilinear form $a$ and a linear form $L$, and the total potential energy takes the form

$$\Pi(u) = \tfrac12\, a(u, u) - L(u) ,$$

i.e. $\Psi(u) = \tfrac12\, a(u, u)$ and $W(u) = L(u)$. After the $s$ integrations by parts of lecture 1, $a$ contains only derivatives of order $s$.

### Why the minimizer solves $Au = f$

Let $u$ solve $Au = f$ and $v$ be admissible. With $L(v) = a(u, v)$ and the symmetry of $a$

$$\Pi(v) = \tfrac12\, a(v, v) - a(u, v) = \tfrac12\, a(v - u, v - u) - \tfrac12\, a(u, u) .$$

Since $a(v - u, v - u) > 0$ for $v \neq u$, $\Pi$ is smallest for $v = u$, with $\Pi(u) = -\tfrac12\, a(u, u) = -\Psi(u)$.

### Directional derivative: the weak form

At the minimum the **directional derivative** vanishes in every admissible direction $v$ (with $v = 0$ on the Dirichlet boundary):

$$\delta\Pi(u; v) := \frac{d}{d\varepsilon}\, \Pi(u + \varepsilon v)\,\Big|_{\varepsilon = 0} = a(u, v) - L(v) = 0 \qquad \text{for all } v .$$

This is the weak form, the variational formulation of lecture 1, with the test function as direction.

### Euler–Lagrange equation: the strong form

Write $\Pi$ as the integral of an energy density $\mathcal{L}$. For $\Pi(u) = \int_0^L \mathcal{L}(x, u, u', u'') \, dx$ the minimizer satisfies the **Euler–Lagrange equation**

$$\frac{\partial \mathcal{L}}{\partial u} - \frac{d}{dx}\frac{\partial \mathcal{L}}{\partial u'}
  + \frac{d^2}{dx^2}\frac{\partial \mathcal{L}}{\partial u''} = 0 ,$$

and for $\Pi(u) = \int_\Omega \mathcal{L}(x, u, \nabla u) \; dx$

$$\frac{\partial \mathcal{L}}{\partial u} - \nabla \cdot \frac{\partial \mathcal{L}}{\partial \nabla u} = 0 .$$

Applied to the total potential energy it gives the **strong form** $Au = f$, of order $2s$ if $\mathcal{L}$ contains derivatives of order $s$:

- **Heat conduction** (lecture 1, $s = 1$): $\mathcal{L} = \tfrac12\, |\nabla u|^2 - f\, u$ with $\partial \mathcal{L} / \partial u = -f$ and $\partial \mathcal{L} / \partial \nabla u = \nabla u$, hence $-f - \nabla \cdot \nabla u = 0$, i.e. $-\nabla \cdot \nabla u = f$.
- **Beam** ($s = 2$): $\mathcal{L} = \tfrac12\, EI\, (u'')^2 - q\, u$ with $\partial \mathcal{L} / \partial u = -q$, $\partial \mathcal{L} / \partial u' = 0$ and $\partial \mathcal{L} / \partial u'' = EI\, u''$, hence $-q + (EI\, u'')'' = 0$, i.e. $EI\, u'''' = q$ for constant $EI$.

Boundary terms of $\Pi$ do not enter the Euler–Lagrange equation; they give the natural boundary conditions (next section).

The three formulations are equivalent:

$$a(u, v) = L(v) \;\; \text{for all } v \quad\overset{\delta\Pi = 0}{\longleftarrow}\quad \min_u \Pi(u)
  \quad\overset{\text{Euler–Lagrange}}{\longrightarrow}\quad Au = f .$$

### Boundary terms: the functional contains the Neumann conditions

Loads on the boundary do work too. A traction $T$ on $\Gamma$ adds the term $-\int_\Gamma T \cdot u \; ds$ to $\Pi$; for heat conduction the term is $+\int_\Gamma g\, u \; ds$ (exercise 1).

- **Natural (Neumann) conditions** follow from the minimization: integrating $\delta\Pi$ by parts leaves boundary terms, which must vanish (beam example below).
- **Essential (Dirichlet) conditions** restrict the admissible functions and are imposed on the function space (`DirichletBC`).

### Example: the biharmonic equation (beam bending)

A beam of length $L$ and bending stiffness $EI$ with deflection $u(x)$, distributed load $q$, clamped at $x = 0$, loaded at the free end by a force $P$ and a moment $M$ (positive in the direction of $u$ and $u'$):

$$\Pi(u) = \underbrace{\int_0^L \tfrac12\, EI\, (u'')^2 \, dx}_{\Psi}
  \;-\; \underbrace{\Big( \int_0^L q\, u \, dx \;+\; P\, u(L) \;+\; M\, u'(L) \Big)}_{W} ,
  \qquad u(0) = 0, \quad u'(0) = 0 .$$

- **Directional derivative** (for all $v$ with $v(0) = v'(0) = 0$), the weak form with $a(u, v) = \int_0^L EI\, u''\, v'' \, dx$ and $s = 2$:
  $$\delta\Pi(u; v) = \int_0^L EI\, u''\, v'' \, dx - \int_0^L q\, v \, dx - P\, v(L) - M\, v'(L) = 0 .$$
- **Integration by parts** (twice):
  $$\int_0^L EI\, u''\, v'' \, dx = \int_0^L EI\, u''''\, v \, dx + EI\, u''(L)\, v'(L) - EI\, u'''(L)\, v(L) ,$$
  hence
  $$\delta\Pi(u; v) = \int_0^L \big(EI\, u'''' - q\big)\, v \, dx + \big(EI\, u''(L) - M\big)\, v'(L)
    - \big(EI\, u'''(L) + P\big)\, v(L) = 0 .$$
- **Strong form:** $EI\, u'''' = q$ in $(0, L)$ with the natural boundary conditions $EI\, u''(L) = M$ (bending moment) and $-EI\, u'''(L) = P$ (shear force) at the free end. The clamping $u(0) = u'(0) = 0$ is essential.

The load vector of exercise 2 in lecture 1, $F_3 = -u'''(1)$ and $F_4 = u''(1)$, is the boundary work $P\, u(1) + M\, u'(1)$.

### From the functional to a system of equations

Inserting $u_h = \sum_i U_i\, h_i$ turns the minimization over functions into the minimization of a function of the $U_i$ (Ritz method):

$$\Pi(U) = \tfrac12\, U^T K\, U - U^T F ,
  \qquad K_{ji} = a(h_i, h_j), \quad F_j = L(h_j) .$$

The condition $\partial \Pi / \partial U_j = 0$ gives $K\, U = F$, the system of the Galerkin method; $K$ is the stiffness matrix, $F$ the load vector, and $\Pi(U)$ is the function minimized by the conjugate gradient method (lecture 2). Since $u_h$ minimizes $\Pi$ over $V_h$:

- $u_h$ is the **best approximation** of $u$ in the energy norm, $a(u - u_h, u - u_h) \le a(u - v_h, u - v_h)$ for all $v_h \in V_h$.
- $a(u_h, u_h) \le a(u, u)$: the approximation is too stiff.

**In FEniCS** the functional is written down directly, `derivative` computes $\delta\Pi(u; v)$, and `solve(dPi == 0, u, bc)` finds the stationary point with Newton's method.

## A gentle introduction to continuum mechanics

### Displacement, strain and strain energy

A material point at $x$ moves to $x + u(x)$ with the **displacement field** $u$. For small deformations the **strain** is

$$\varepsilon(u) = \tfrac12 \left(\nabla u + \nabla u^T\right), \qquad
  \varepsilon_{ij} = \tfrac12 \left(\frac{\partial u_i}{\partial x_j} + \frac{\partial u_j}{\partial x_i}\right) .$$

Its trace $\operatorname{tr}\varepsilon = \nabla \cdot u$ is the relative volume change, the **deviator** $\operatorname{dev}\varepsilon = \varepsilon - \tfrac13\, (\operatorname{tr}\varepsilon)\, I$ is the change of shape.

The **strain energy density** of a linear isotropic material, with the **bulk modulus** $\kappa$ and the **shear modulus** $G$:

$$\psi(\varepsilon) = \tfrac12\, \kappa\, (\operatorname{tr}\varepsilon)^2 + G\, \operatorname{dev}\varepsilon : \operatorname{dev}\varepsilon ,
  \qquad A : B = \sum_{i, j} A_{ij}\, B_{ij} .$$

The **stress** is the derivative of the strain energy density with respect to the strain,

$$\sigma = \frac{\partial \psi}{\partial \varepsilon} = \kappa\, (\operatorname{tr}\varepsilon)\, I + 2G\, \operatorname{dev}\varepsilon ,$$

and $\sigma\, n$ is the force per unit area on a surface with normal $n$. With the **Lamé constants** $\lambda = \kappa - \tfrac23 G$, $\mu = G$ this is Hooke's law $\sigma = \lambda\, (\operatorname{tr}\varepsilon)\, I + 2\mu\, \varepsilon$ (exercise 3).

### Measuring the constants

In a **tension test** a rod of cross section $S$ is pulled by the force $F$; one measures the elongation $\Delta L$ and the shrinkage $\Delta d$ of the width. **Young's modulus** $E$ and **Poisson's ratio** $\nu$ are

$$E = \frac{F / S}{\Delta L / L}, \qquad \nu = -\frac{\Delta d / d}{\Delta L / L} .$$

They are related to $\kappa$ and $G$ by (exercise 3)

$$\kappa = \frac{E}{3\, (1 - 2\nu)}, \qquad G = \frac{E}{2\, (1 + \nu)}, \qquad\qquad
  E = \frac{9\, \kappa G}{3\kappa + G}, \qquad \nu = \frac{3\kappa - 2G}{2\, (3\kappa + G)} .$$

### The total potential energy of an elastic body

$$\Pi(u) = \int_\Omega \psi(\varepsilon(u)) \; dx - \int_\Omega f \cdot u \; dx - \int_\Gamma T \cdot u \; ds$$

with the body force $f$ (own weight: $f = (0, -\rho g)$) and the **traction** $T$ on the Neumann part $\Gamma$.

- **Directional derivative:** with $\delta\psi = \sigma : \varepsilon(v)$,
  $$\delta\Pi(u; v) = \int_\Omega \sigma(u) : \varepsilon(v) \; dx - \int_\Omega f \cdot v \; dx - \int_\Gamma T \cdot v \; ds = 0
    \qquad \text{for all } v \text{ with } v = 0 \text{ on the Dirichlet boundary} .$$
- **Integration by parts** (divergence theorem, $\sigma$ symmetric):
  $\int_\Omega \sigma : \varepsilon(v) \; dx = -\int_\Omega (\nabla \cdot \sigma) \cdot v \; dx + \int_{\partial\Omega} (\sigma n) \cdot v \; ds$, hence
  $$\delta\Pi(u; v) = -\int_\Omega \big(\nabla \cdot \sigma + f\big) \cdot v \; dx + \int_\Gamma \big(\sigma n - T\big) \cdot v \; ds = 0 .$$
- **Strong form:** $-\nabla \cdot \sigma(u) = f$ in $\Omega$ with the natural boundary condition $\sigma n = T$ on $\Gamma$ ($T = 0$ on a free surface).

**Plane strain:** $u_z = 0$ and no variation in $z$, so $\varepsilon_{zz} = 0$. The deviator is taken in 3D: $\operatorname{dev}\varepsilon : \operatorname{dev}\varepsilon = \varepsilon : \varepsilon - \tfrac13\, (\operatorname{tr}\varepsilon)^2$, as in `psi` below.

## Examples in FEniCS: everything as a minimization problem

Write down the functional, let `derivative` produce the variational formulation, solve. All examples are 2D linear elasticity on the same beam with holes.

```python
from fenics import *
from matplotlib import pyplot
import numpy as np
%matplotlib inline

set_log_level(LogLevel.WARNING)   # silence the solver messages
```

### A beam with holes, without a mesh generator

The beam $\Omega = (0, 5) \times (0, 1)$ with four holes of radius $R = 0.3$ at $x = 1, 2, 3, 4$. `SubMesh` cuts a square of half-width $a = 0.25$ around each hole centre out of a crossed grid with $h = 0.05$. Then every point $X$ in the unit square around a hole moves along the ray from the hole centre: with $d = X - \text{centre}$, the distance $s = \max(|d_1|, |d_2|)$ in the maximum norm and $t = (s - a) / (0.5 - a)$, its new distance from the centre is $(1 - t)\, R + t\, r_{\text{edge}}$, where $r_{\text{edge}} = 0.5\, |d| / s$ is the distance of the edge of the unit square along the same ray. The square hole becomes the circle of radius $R$; the edges of the unit squares stay, so neighbouring squares fit together.

```python
length, height = 5.0, 1.0
R, a = 0.3, 0.25     # hole radius, half-width of the square cut out

box  = RectangleMesh(Point(0.0, 0.0), Point(length, height),
                     100, 20, 'crossed')
keep = MeshFunction('size_t', box, 2, 1)
for c in cells(box):     # cut out a square around each hole
    x, y = c.midpoint().x(), c.midpoint().y()
    if 0.5 < x < 4.5 and max(abs(x - round(x)), abs(y - 0.5)) < a:
        keep[c] = 0
mesh = SubMesh(box, keep, 1)

for X in mesh.coordinates():     # map the squares onto circles
    if 0.5 < X[0] < 4.5:
        centre = np.array([round(X[0]), 0.5])
        d = X - centre
        s, r = abs(d).max(), np.hypot(*d)
        t = (s - a)/(0.5 - a)    # 0 on the hole, 1 on the cell edge
        X[:] = centre + d*((1 - t)*R/r + t*0.5/s)

print("%d triangles, %d nodes" % (mesh.num_cells(), mesh.num_vertices()))
pyplot.figure(figsize=(10, 2.4), dpi=100)
plot(mesh, linewidth=0.3)
pyplot.savefig("beam-mesh.png", bbox_inches='tight')
```

### Linear elasticity in 2D: shared setup

The material of all examples, $E = 1000$ and $\nu = 0.3$, converted to the bulk and the shear modulus; plane strain as above. `ds(1)` integrates over the right end only. `show` draws the deformed beam at true scale (dashed: before), each triangle shaded by its strain energy density on a square-root scale.

```python
E, nu = 1000.0, 0.3
rho_g = 0.3
kappa = E/(3*(1 - 2*nu))   # bulk modulus
G     = E/(2*(1 + nu))     # shear modulus

V = VectorFunctionSpace(mesh, 'Lagrange', 1)

def epsilon(u):
    return sym(grad(u))

def psi(eps):              # strain energy density, plane strain
    return 0.5*kappa*tr(eps)**2 + G*(inner(eps, eps) - tr(eps)**2/3)

tol = 1e-3
def left(x, on_boundary):
    return on_boundary and near(x[0], 0.0, tol)

def right(x, on_boundary):
    return on_boundary and near(x[0], length, tol)

boundaries = MeshFunction('size_t', mesh, 1, 0)
AutoSubDomain(right).mark(boundaries, 1)       # ds(1): right end
ds = Measure('ds', domain=mesh, subdomain_data=boundaries)

# plotting: the deformed beam, each triangle shaded by its strain energy density
from matplotlib.collections import LineCollection
from matplotlib.colors import PowerNorm
from matplotlib.tri import Triangulation

DG0   = FunctionSpace(mesh, 'DG', 0)
cell_dof = DG0.dofmap().entity_dofs(mesh, 2)   # cell -> its value in DG0
bmesh = BoundaryMesh(mesh, 'exterior')
bnode = bmesh.entity_map(0).array()            # boundary node -> mesh node

def show(u, filename):
    X = mesh.coordinates()
    Y = X + u.compute_vertex_values(mesh).reshape(2, -1).T     # deformed nodes
    energy = project(psi(epsilon(u)), DG0).vector().get_local()[cell_dof]
    pyplot.figure(figsize=(10, 2.6), dpi=100)
    c = pyplot.tripcolor(Triangulation(Y[:, 0], Y[:, 1], mesh.cells()), facecolors=energy, cmap='Blues',
                         norm=PowerNorm(0.5, vmin=0.0, vmax=np.percentile(energy, 99)))
    edges = bmesh.cells()
    pyplot.gca().add_collection(LineCollection(X[bnode][edges], colors='0.6', linewidths=0.8, linestyles='dashed'))
    pyplot.gca().add_collection(LineCollection(Y[bnode][edges], colors='0.3', linewidths=0.8))
    pyplot.colorbar(c, fraction=0.03, pad=0.02, shrink=0.8, extend='max', label='strain energy density')
    pyplot.gca().set_aspect('equal')
    pyplot.axis('off')
    pyplot.savefig(filename, bbox_inches='tight')
```

### 1. Dirichlet – Dirichlet: a tension test

Left end fixed, $u = (0, 0)$, right end pulled to $u = (0.25, 0)$; no body force and no traction, so $\Pi(u) = \int_\Omega \psi(\varepsilon(u)) \; dx$. The printed stored energy is the work needed to pull the end. The same code on the grid without holes gives 6.94: the holes remove 22 % of the material but halve the stiffness. The energy concentrates in the thin bands above and below the holes.

```python
bc0 = DirichletBC(V, Constant((0.0, 0.0)), left)
bc1 = DirichletBC(V, Constant((0.25, 0.0)), right)

u = Function(V)
v = TestFunction(V)

Pi  = psi(epsilon(u)) * dx
dPi = derivative(Pi, u, v)
solve(dPi == 0, u, [bc0, bc1])

print("stored strain energy: %.4f" % assemble(Pi))
show(u, "elasticity-0.png")
```

### 2. Dirichlet – homogeneous Neumann: own weight

Left end clamped, body force $f = (0, -\rho g)$ with $\rho g = 0.3$. The functional has no boundary term; the minimization delivers $\sigma n = 0$ on the rest of the boundary, the holes included, without any code. Check $\Pi(u) = -\Psi(u)$, as for the spring. Without the holes the tip deflection is $-0.265$: the holes sit at the neutral axis, they remove 22 % of the weight but little bending stiffness.

```python
bc = DirichletBC(V, Constant((0.0, 0.0)), left)
f  = Constant((0.0, -rho_g))

u = Function(V)
v = TestFunction(V)

Pi  = (psi(epsilon(u)) - dot(f, u)) * dx
dPi = derivative(Pi, u, v)
solve(dPi == 0, u, bc)

print("tip deflection: %.4f" % u(length, 0.5)[1])
print("Pi(u) = %.4f,  -Psi(u) = %.4f" % (assemble(Pi), -assemble(psi(epsilon(u))*dx)))
show(u, "elasticity-1.png")
```

### 3. Dirichlet – Neumann: an end load

Left end clamped, no body force, traction $T = (0, -0.5)$ on the right end $\Gamma$. Its work is the boundary term of the functional, `ds(1)` restricts the integral to the right end, and the natural condition $\sigma n = T$ comes with it. Without the holes the tip deflection is $-0.233$: under the same load the beam with holes is 21 % softer.

```python
T = Constant((0.0, -0.5))

u = Function(V)
v = TestFunction(V)

Pi  = psi(epsilon(u)) * dx - dot(T, u) * ds(1)
dPi = derivative(Pi, u, v)
solve(dPi == 0, u, bc)

print("tip deflection: %.4f" % u(length, 0.5)[1])
print("Pi(u) = %.4f,  -Psi(u) = %.4f" % (assemble(Pi), -assemble(psi(epsilon(u))*dx)))
show(u, "elasticity-2.png")
```
